```python
# Exemplo de criação do DataFrame df_novos_dados que será utilizado no merge
from pyspark.sql import Row

# Dados de exemplo contendo cliente_id e demais colunas relevantes
novos_dados = [
    Row(cliente_id=1, nome='Lucas', email='lucas@email.com', idade=28),
    Row(cliente_id=2, nome='Ana', email='ana@email.com', idade=31)
]

df_novos_dados = spark.createDataFrame(novos_dados)

df_novos_dados.show()  # Visualização dos dados para conferência
```

```python
%sql
Select *
from workspace.meu_database.clientes
```

```python
# Supondo que você tenha um DataFrame chamado df_novos_dados para fazer o merge
# Ajuste 'cliente_id' para a chave primária correta da tabela

from delta.tables import DeltaTable

delta_table = DeltaTable.forName(spark, "workspace.meu_database.clientes")

delta_table.alias("clientes").merge(
    df_novos_dados.alias("novos"),
    "clientes.cliente_id = novos.cliente_id"
).whenMatchedUpdateAll() \
 .whenNotMatchedInsertAll() \
 .execute()
```
